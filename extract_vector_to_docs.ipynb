```python
import json
from langchain.vectorstores.chroma import Chroma
from get_embedding_function import get_embedding_function
 
# Load Chroma DB
CHROMA_PATH = "chroma"
embedding_function = get_embedding_function()
db = Chroma(persist_directory=CHROMA_PATH, embedding_function=embedding_function)
 
# Fetch raw data from DB
raw_data = db.get()
 
# Prepare data for JSON
json_data = []
for doc, metadata, doc_id in zip(raw_data["documents"], raw_data["metadatas"], raw_data["ids"]):
    json_data.append({
        "id": doc_id,
        "content": doc,
        "metadata": metadata
    })
 
# Save to JSON file
with open("chroma_documents.json", "w", encoding="utf-8") as f:
    json.dump(json_data, f, ensure_ascii=False, indent=2)
 
print("Documents exported to chroma_documents.json")
```
